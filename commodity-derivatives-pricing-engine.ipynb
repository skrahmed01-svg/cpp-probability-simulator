```python
%%writefile monte_carlo.cpp
#include <iostream>
#include <vector>
#include <cmath>
#include <random>
#include <chrono>

int main() {
    // Parameters
    const int num_simulations = 1000000;
    const double S0 = 1.20;
    const double r = 0.02;
    const double sigma = 0.25;
    const double T = 0.5;

    // Random Number Generation
    std::mt19937_64 rng(42); // Standard 64-bit Mersenne Twister engine seeded with 42
    std::normal_distribution<double> dist(0.0, 1.0);

    double sum_prices = 0.0;

    auto start = std::chrono::high_resolution_clock::now();

    // 1,000,000 Single-Step Monte Carlo Runs
    double drift = (r - 0.5 * sigma * sigma) * T;
    double vol = sigma * std::sqrt(T);

      for (int i = 0; i < num_simulations; ++i) {
        double z = dist(rng);
        double S_T = S0 * std::exp(drift + vol * z);
        sum_prices += S_T;
    }

    double avg_price = sum_prices / num_simulations;

    auto end = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double> diff = end - start;

    std::cout << "Average Simulated Price: $" << avg_price << std::endl;
    std::cout << "Execution Time: " << diff.count() << " seconds" << std::endl;

    return 0;
}
```
